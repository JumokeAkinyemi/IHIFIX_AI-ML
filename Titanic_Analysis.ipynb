DETECT OUTLIERS

```python
# Calculate Q1 (25th percentile) and Q3 (75th percentile) for the 'Fare' column
Q1 = titanic_data['Fare'].quantile(0.25)
Q3 = titanic_data['Fare'].quantile(0.75)

# Calculate the Interquartile Range (IQR)
IQR = Q3 - Q1

# Define the upper and lower bounds for outlier detection
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers = titanic_data[(titanic_data['Fare'] < lower_bound) | (titanic_data['Fare'] > upper_bound)]

print(f"Number of outliers in 'Fare' column: {len(outliers)}")
print("Outliers:")
print(outliers[['Fare']])
```

```python
# Cap the outliers in the 'Fare' column
titanic_data['Fare'] = np.where(titanic_data['Fare'] < lower_bound, lower_bound, titanic_data['Fare'])
titanic_data['Fare'] = np.where(titanic_data['Fare'] > upper_bound, upper_bound, titanic_data['Fare'])

print("Outliers in 'Fare' column have been capped.")

# Verify by checking the new boxplot
sns.boxplot(x=titanic_data['Fare'])
plt.title('Fare Distribution After Outlier Capping')
plt.show()
```
